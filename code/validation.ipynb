```python
import pandas as pd
from IPython.display import display
```

```python
# load file csv-nya
df = pd.read_csv('../datasets/data.csv', index_col=0)
```

```python
# cek apakah data berupa numerik
numeric_columns = df.select_dtypes(include=['number']).shape[1]
total_columns = df.shape[1]

if numeric_columns == total_columns:
    data_type = "Numeric"
else:
    error_columns = total_columns - numeric_columns
    data_type = f"Ada Error! ({numeric_columns} Numeric, {error_columns} Non-Numeric)"
```

```python
# validasi jumlah kolom dan sampel
summary = pd.DataFrame({
    "Dimensi": ["Rows (Samples)", "Columns (Features)", "Data Type"],
    "Data": [df.shape[0], df.shape[1], data_type]
})
```

```python
# hasil harus berupa numerik
display(summary)
```

Hasil validasi menunjukkan terdapat 801 sampel dan 20531 fitur serta data yang dimiliki semuanya berupa numerik. Ini menandakan dataset ini bisa digunakan untuk kebutuhan topik skripsi yang saya angkat saat ini mengingat algoritma K-means bekerja dengan sangat baik dengan data numerik karena algortima K-means menggunakan Euclidean Distance untuk mengukur jarak antar data.
